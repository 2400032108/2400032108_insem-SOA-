```python
students = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
    "David": 88
}


highest_scorer = max(students, key=students.get)
highest_marks = students[highest_scorer]


print("Student with the highest marks:")
print(f"{highest_scorer} scored {highest_marks}")
```
